
# Regular Expression: Essential Match Search and Findall

The regular expression is a special character sequence, checking whether a string matches a pattern. Python had `re`  since version 1.5. Regular expressions are a huge topic; many books were written just on the topic. It is hard to do justice within just a single subsection. 

#### Raw String 'r'
* The raw type string is created by adding the character `'r'` before quotation mark.
* When a string is of `the raw type`, the compiler does not make any substitution for it. 
* Comparing Python `normal string` vs `raw string` 

```python
# Define a normal python string.
normal_string = 'I am a\nnormal string'

# Add r in front of a python string will convert it to a raw type string.
raw_string = r'I am a\nnormal string' + " with raw"

# The escape character \n in the normal string will take effect.
print(normal_string)


# The \n does not take effect in the python raw type string. 
print(raw_string)

```

```python
print('a\tb\tc')
```

```python
print(r'a\tb\tc')
```

It is good practice to use a raw string.

## Match

#### The `match(pat, str)` method can find the matching object `only when the beginning `of the searched string matches the pattern.

```python
import re
 
# The pattern is 'dog', the searched string is 'dog cat dog' which start with 'dog', 
# so the match function return an match object.
re.match(r'dog', 'dog cat dog')
```

```python
re.match(r'cake', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake, Jedi said')
```

##### match group() 

```python
 
match = re.match(r'dog', 'dog cat dog')
 
match.group(0)
```

```python
matchme =re.match(r'cake', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake, Jedi said')
```

```python
matchme.group(0)
```

```python
type(matchme.group())
```

```python
 
# The pattern is 'dog', the searched string is 'cat dog' which does 
#not start with 'dog', so the match function returns None. 
re.match(r'dog', 'cat dog')
```

```python
match = re.match(r'dog', 'cat dog')
 
print(match)
```

```python
import re
matchme =re.match(r'cake', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
# The match object's start() method return the beginning index number of the match. 
matchme.start()
```

```python
# The match object's end() method return the ending index number of the match.  
matchme.end()
```

## Search() method

#### The `search(pat, str)` method is similar to match(), but the search() method looks for all positions

```python
import re
 
searchme = re.search(r'dog', 'cat dog')
```

```python
print(searchme)
```

```python
searchme.group(0)
```

```python
searchme2 =re.search(r'eat', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
searchme2
```

#### But, the search() method will `stop searching` after `it finds a match`,
#### the search() method`only find its first occurrence`

```python
import re

searchdog = re.search(r'dog', 'cat dog and dog')
 
# Only return the first match.
searchdog.group(0)
```

```python
searchdog.group(1)
```

```python
searchdog.group()
```

```python
type(searchdog.group())
```

## Findall()

#### the `findall()` method can  get a list of `all` matched strings

```python
import re
re.findall(r'eat', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
eats=re.findall(r'eat', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
type(eats)
```

```python
eats=re.findall(r'carrotcake', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
type(eats)
```

```python
eats
```

## Special Character . (period); ^ caret; \$ dollar; 

#### Period . Matches any single character unless the newline character

```python
re.search(r'ca.e', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake').group(0)
## only in the beginning
```

```python
re.findall(r'ca.e', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

#### .. find matches for two charaters; ... three characters

```python
re.findall(r"...", "abc12ab3")
```

```python
re.findall(r"...", "a\nbcde")
```

```python
re.findall(r"...", r"a\nbcde")
```

#### A caret ^. Matches the`start` of the string.

```python
re.search(r'^cake', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake').group()
```

```python
## However, the code below will NOT give the result. Try it for yourself:
re.search(r'^eat', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake').group()
```

```python
import re

result = re.search(r'^eat', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')

if result:
    print(result.group())
else:
    print("No match found.")
```

#### dollar \$ - Matches the `end` of string.

```python
re.search(r'cake$', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake')
```

```python
re.search(r'cake$', 'cake I eat; carrot I eat also; of course, I also eat a carrotcake').group()
```

```python
re.findall(r'n$', "n")
```

```python
re.findall(r'n$', "question")
```

```python
re.findall(r'n$', "questionnaire")
```
