## 4.002 Exploring the CSV data format






What is CSV?

- CSV stands for comma-separated values, meaning that the data is delimited. 
- How we might want to import the data, so here we've got some options. For instance, to begin importing from row one. 
- We can see the data can separated by three different delimiters: **Tab-based delimiters, comma, and Semicolon**


Here we have some structure in the data, and we can begin to see that the data has some meaning attached to it:
- So we have location, indicator, subject, measure, frequency, time, value. We also call them as "columns, fields"

We can begin to see where the headings emerge at the top, and we can also see that the data is in fact delimited by commas.

## 4.005 A re-introduction to JSON

Differences between JSON and CSV
- CSV data in data frames both have rows and columns and therefore rectangular in shape. 
  - The columns are the different elements of the data, and they define things that are interesting to us. 
  - The rows represent each item in a data set with specific values for each of the elements. 
  
- JSON data is most commonly encountered when accessing live data from the Internet. 
  
Many websites provide two ways to view the content: 
1. You can view the content via a normal web browser interface that most of us use the same, 
2. but we can also look more directly at the data itself or through the API. 

**Internet data services often provide their data in JSON format.**
  -JSON data has a tree structure instead of a rectangular structure,
  
For instance, we may have a field in JSON that defines a member of staff, who has a number of things associated with them. 
```json
{
"staff":{"name":"Sean", "age":31, "city":"London"}   
}
```
- In the previous examples, we looked at columns and rows, defining that data and defining the structure within the data. - - In Json, we have a member of staff, which is generic class that we've defined. They have a name object, in this case, "My name is Sean". Their age is defined and the city in which they live is defined. 

These **curly braces** define levels by which an index is required. When we talks about the structure, what we're really looking at is this tree-based structure where we have an object belonging `to greater objects or higher object up` in **hierarchy**

````python


a = np.array([1,2,3,4,5])
l = len(a)

mid = l//2 if l%2 == 0 else l//2+1

for i in range(mid):
    if (i+mid) < l:
        a[i], a[mid+i] = a[mid+i], a[i]

        
print(a)
````

1. **Midpoint Calculation**:
   ```python
   mid = l // 2 if l % 2 == 0 else l // 2 + 1
   ```
   This calculates the midpoint of the array. If the length `l` is even, `mid` is `l // 2`. If `l` is odd, `mid` is `l // 2 + 1`. In this case, since `l` is `5` (odd), `mid` is `3`.

4. **Swapping Elements**:
   ```python
   for i in range(mid):
       if (i + mid) < l:
           a[i], a[mid + i] = a[mid + i], a[i]
   ```
   This loop iterates from `0` to `mid - 1` (i.e., `0` to `2` in this case). For each `i`, it swaps the element at index `i` with the element at index `mid + i`, provided `mid + i` is within the bounds of the array.

   - **Iteration 1** (`i = 0`):
     - Swap `a[0]` and `a[3]` → `[4, 2, 3, 1, 5]`
   - **Iteration 2** (`i = 1`):
     - Swap `a[1]` and `a[4]` → `[4, 5, 3, 1, 2]`
   - **Iteration 3** (`i = 2`):
     - No swap since `i + mid` (i.e., `5`) is not less than `l` (i.e., `5`).

5. **Print Result**:
   ```python
   print(a)
   ```
   This prints the modified array: `[4, 5, 3, 1, 2]`.

So, the code effectively swaps the first half of the array with the second half, but only up to the midpoint. The result is `[4, 5, 3, 1, 2]`.

### Data Provenance

Data provenance is the documentation of where a piece of data comes from and the processes and methodology by which it was produced. The word provenance originates from the French term 'provenir' meaning 'to come from' and is also known as 'lineage' or 'pedigree'. "Lineage" and "pedigree" are terms often used to describe the ancestry, heritage, or chronological sequence of events in a particular order.

https://faculty.washington.edu/hazeline/ProvEco/generic2.html

https://ardc.edu.au/resource/data-provenance/

Data provenance tells how data look like at all stages of the analysis process.

To maintain data provenance:
* Keep a copy of the raw data,
* Keep a log (step/code/instruction) of what kind of cleaning transformations were applied
(with the aim to be `reproducable`) 
* Keep a copy of the cleaned or pristine data.

Both **data lineage** and **data provenance** deal with **transformation** and **processing** — and this is one of the reasons why people often get confused about the distinction between the two. Both concepts track the stages and activities around data, but they serve slightly different **goals** and **focuses**.


### **Data Lineage** and Transformation/Processing:
- **Data Lineage** **does** track data transformation and processing. It focuses on the **flow of data** across different systems and the **transformations** it undergoes.
- **Transformations** could be things like data cleaning, combining data from multiple sources, applying business logic, aggregating, etc.
- So, when data is processed or transformed, **data lineage** will show how this data moves and changes over time, from the source to the destination.

### **Data Provenance** and Transformation/Processing:
- **Data Provenance**, on the other hand, focuses on **understanding the origin and history of data**, with a more detailed look at the **steps involved** in how the data came to be at each stage.
- **Processing** and **transformation** are still part of **data provenance**, but it's specifically concerned with the **authenticity** and **trustworthiness** of the data at each stage. 
- For example, data provenance will keep track of **what exactly happened** to the data — who processed it, which tools were used, and when it was processed, giving a deeper **audit trail** of actions around the data.

### Key Differences in Relation to Transformation/Processing:

- **Data Lineage** is more about **tracking the flow** of data and how it moves through various systems, getting transformed, aggregated, and eventually stored or used.
- **Data Provenance** is more about **understanding the history and origin** of the data, including **who or what** made changes, and how transformations affect its credibility or authenticity.

### Both Track Transformation and Processing, but Differ in Focus:
- **Data Lineage** will show **where the data went** and **what transformations** occurred along the way (such as from raw data to cleaned data).
- **Data Provenance** will show **how the data was created**, **who touched it**, and **how its transformations** at each stage affect its reliability (e.g., whether a change was made correctly, or by an authorized process).

### To summarize:
- **Data Lineage**: Tracks **data flow** and transformations across the entire lifecycle of the data.
- **Data Provenance**: Tracks **data origins** and gives more **detailed information about the processes** that affect its trustworthiness, including transformations.

Both concepts **do track transformations and processes**, but **data lineage** is more about the **movement and flow** of data, whereas **data provenance** is about the **origin and integrity** of data, including those processes.
