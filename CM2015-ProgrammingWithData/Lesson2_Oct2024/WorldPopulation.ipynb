Your next job is to try to write a program! Here's the brief.

Write a program that calculates the size of the world's population at the start of 2022.

The population at the start of 2018 was 7.6 billion. The growth rate at the time was 1.09% per year.  

Assume a fixed population growth rate over the time period.
Put the starting population and growth rate into variables.

````python
# Constants
population_2018 = 7.6  # in billions
growth_rate = 1.09  # annual growth rate in percentage
````

Carry out some mathematical operations to compute the population at the start of 2022.
Print out the result.

````python
# Calculate population at the start of 2022
population_2022 = population_2018 * (1 + growth_rate / 100) ** 4

# Print the result
print(f"The estimated world population at the start of 2022 is approximately {population_2022} billion.")
````

````python
print(f"The estimated world population at the start of 2022 is approximately {population_2022:.2f} billion.")
````

```
The ** operator is used for exponentiation in Python. In the context of calculating population growth, raising a value to the power of 4 corresponds to a 4-year period.

Here's the rationale:

The population growth rate is given as an annual percentage, and you want to calculate the population at the start of 2022, which is four years after the start of 2018.

To calculate the population after four years of growth, you need to raise the factor (1 + growth_rate / 100) to the power of 4. This represents compounding the growth over four years.

For example, if the growth rate is 1.09% per year, then (1 + 1.09/100) represents the growth factor for one year. Raising it to the power of 4 accounts for the growth over four years. This is a simplified model that assumes a fixed growth rate over the four-year period, which may not be entirely accurate but serves as a basic approximation.
```

Challenge exercise: can you convert your code into a function that takes as its inputs the population growth rate, the population in 2018 and the year we would like to know the population in. e.g.:

def get_population(growth_rate, pop_in_2018, target_year)

It should return the population in target_year.

````python



# Challenge exercise - Function to calculate population for any year
def get_population(growth_rate, pop_in_2018, target_year):
    years_elapsed = target_year - 2018
    return pop_in_2018 * (1 + growth_rate / 100) ** years_elapsed

# Test the function
target_year = 2022
population_target = get_population(growth_rate, population_2018, target_year)
print(f"The estimated world population at the start of {target_year} is approximately {population_target:.2f} billion.")
````
