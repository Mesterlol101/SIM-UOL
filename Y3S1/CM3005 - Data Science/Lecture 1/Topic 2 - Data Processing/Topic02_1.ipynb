### 2.1.1 Advanced Data Processing Using NumPy

#### Data Preprocessing
* Machine learning, data visualization, computer vision and other data science-related fields require datasets to be consistent and well-structured
* Common issues with data are:
    - Missing values
    - Type inconsistency or type mismatch
    - Duplicating values, including entire rows and columns
* Resolving such problems is the first step in many data science tasks, and is known as ‘preprocessing’
    - Corresponds to bringing a database to first normal form (1NF)
* Common preprocessing task is to extract useful data from an existing dataset
    - This amounts to generating a sub-array based on an original array, and can be done with NumPy array using advanced indexing. 

#### Advanced Indexing

* Advanced indexing allows us to build new arrays from existing ones by specifying a set of elements to include
  - Can be applied to arrays of different dimensions  
  - Code will show advanced indexing with 1D and 2D arrays  
* In most cases, data science deals with data organized in table-like structures, similar to XML files or spreadsheets  
* It's common for such tables to contain rows or columns that hinder further processing — often due to missing values, data type mismatches, or other inconsistencies  
* Removing (or "dropping") problematic rows and columns is usually the first, and sometimes the only, step in preprocessing  
* This can also be accomplished using other libraries (e.g., pandas) — which offer alternative methods that produce similar results

#### Structured Data

* Datasets can be split into two groups, based on the type of their data points
  1. Datasets containing data points of the same type, usually numerical, either integer or real
  2. Datasets containing data points of multiple data types, e.g., numerical, textual, Boolean
* NumPy library provides the functionality to store and process arrays of structured data, i.e., data consisting of multiple data types
  - The structure needs to be consistent throughout the datasets
  - NumPy array constructed in this way correspond to tables from a relational database


Let's consider the following example -- By using NumPy arrays, organise the information about a number of books with the following information for each item: Title, Year Price

#### Statistics with NumPy

* Statistics is a major branch of mathematics
    - Studies the properties of numerical data; more specifically, data points organised in datasets
    - Can generalise the properties of datasets and draw conclusions based on the data
* This section focuses on the statistical functionality of the NumPy library.
    - More detailed view on statistics in Topics 4 and 5
* Statistics uses two main measurements to describe a datasets
    - measures of central tendency
    - measures of spread
* Measures of central tendency evaluate a central value, around which the data tend to cluster. Popular measures of central tendency are:
    - Mean (also known as the average, or arithmetic average)
    - Median (splits the dataset into two halves)
    - Mode (the data point which appears most frequently)
* Measures of spread evaluate how data are spread around certain central values. Three common measures of spread are:
    - Range
    - Standard Deviation ($sd$)
    - Variance (= $sd^2$)
