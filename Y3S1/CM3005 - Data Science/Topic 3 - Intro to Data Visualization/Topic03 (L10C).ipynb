### Data

- Quantitative: measurements of quantity
- Qualitative: measurements of quality

### DRIP Syndrome

- **Data Rich** but **Information Poor**

### How Do We Make Sense of Data? Data -> Information

- Visualisation: using graphics to explore and understand data (use human visual capacities)
- Data mining: searching for patterns and trends
- Machine learning: build models, test theories, predict the future

### Data Visualisation 

- **Amplify cognition**: Process of making visible the patterns and structures in data using graphs, tables, diagrams, maps, etc.
- *"The representation and presentation of data that exploits our visual perception abilities in order to amplify coginition" (Kirk 2012, p.17)*


<div style="background:#e7f3fe; padding:10px; border-radius:5px;">
  <h2> 3.1 Information and Data </h2>
</div>

### Information Visualisation
- *... is the use of computer-supported, interactive, visual representations of abstract data to amplify cognition (Card et al, 1999, p.7)*
- Quantitative data representing abstract concepts, processes and behaviours, e.g., share prices, child literacy, sales, happinness...
- Focus on understanding social, pychological, economic phenomena

  

  Note: Coursera videos include additional examples

### Scientific Visualisation
- Quantitative data representing physical objects, e.g., geographic data, human body, cell structure, gravity...
- Understanding the structure of the physical world

   
  

  Note: Coursera videos include additional examples

### Information vs Scientific Visualisation
- Similarities
    * use visual modality
    * represent data in ways our brain can comprehend
- Shared objectives
    * amplify cognition
    * create and communicate knowledge

  <h2> 3.2 Visual Perception and Design </h2>
</div>

### The Importance of Vision

*Visual displays provide the highest bandwidth channel from the computer to the human. Indeed, we acquire more information through vision than through all of the other senses combined. (Ware 2012, p.2)*

### Semiotics of Graphics
- Study of visual symbols and how they convey meaning
- Classical view (Saussure)
    * all symbols are arbitrary
    * must be learnt
    * no systems of representation can be better than any other
- Scientific view
    * visual system has particular properties (product of evolution)
    * different forms of representation are more closely aligned with our perception mechanisms
    * some grounds for determining better/worse forms of visual communication

### Sensory vs Arbitrary Symbols
Ware (2012, p.9)
- **sensory** "aspects of visualisations that derive their expressive power from their ability to use the perceptual processing power of the brain *without learning*"
- **arbitrary** "aspects of representation that must be learned, because the representations have no perceptual basis"


#### 1. Sensory Symbols
- Well matched to the early stages of neural processing
- Tend to be stable across individuals, cultures, and time



#### 2. Arbitrary Symbols
- Derive power from culture
- Dependent on individual cultural knowledge

  <h2> 3.3 Purpose and Value of Visualisation </h2>
</div>

### What Can We Do With Visualisations?

'\[Data visualisation is\] an umbrella term to cover all types of visual representations that support the exploration, examination, and communication of data' -- *Few (2009)*

### Data Visualisation Types

Exploratory
- early stages of research
- interactive
- quick and dirty
- helps discover trends and patterns

Explanatory
- late stages of research
- polished, publication/presentation ready
- conveys a clear message
- makes a point or answers a question

### Uses for Data Visualisation

- Analysis
- Communication
- Monitoring
- Planning



### Value of Data Visualisation

'The greatest value of a picture is when it forces us to notice what we never expected to see.' -- *Tukey (1977)*



Note: Coursera videos include additional examples

### Further Values of Data Visualisation
- Industry demand for data analysis skills
- Useful in many areas of work and study
- Practical data manipulation skills
- Experience of research
- Visualisation is a powerful means of communication and making a difference

### Different Ways of Working with Matplotlib

1. pyplot API
    - quick and simple
    - flat namespace
    - automatic state management
    - less fine-grained control
2. object-oriented API
    - modular structure
    - can feel more complex to Python beginners
    - explicit state management
    - full control
3. pandas.plot API
    - data-oriented API
    - quick and simple
    - returns Matplotlib objects
    - full customisation available via object-oriented API

The `matplotlib.pyplot` module contains functions that allow you to generate many kinds of plots quickly. Use the object-oriented approach for more control and customization of your plots.

```python
# Common imports for data analysis
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data
n = 100
df = pd.DataFrame({
    'x':np.random.normal(0, 1, n),
    'y':np.random.normal(0, 2, n)
})
df.head()
```

#### *1. pyplot API*

```python
### Insert code to plot histogram
```

```python
plt.hist(df['x'])
```

```python
plt.hist(df['x'])
plt.show()
```

```python
plt.hist(df['x']);
```

```python
plt.figure()
plt.subplot(1, 2, 1)
plt.hist(df['x'])
plt.show()
```

```python
plt.figure()

plt.subplot(1, 2, 1)
plt.hist(df['x'])

plt.subplot(1, 2, 2)
plt.hist(df['y'])

plt.show()
```

```python
plt.figure()

plt.subplot(1, 2, 1)
plt.hist(df['x'])
plt.title('x')
plt.xlim(-6,6)
plt.ylim(0,25)

plt.subplot(1, 2, 2)
plt.hist(df['y'])
plt.title('y')
plt.xlim(-6,6)
plt.ylim(0,25)

plt.show()
```

#### *2. Oriented-oriented API*

```python
### Insert code to plot histogram
```

```python
df.head()
```

```python
fig, axes = plt.subplots(1, 2, sharex=True, sharey=True)
axes[0].hist(df['x'])
axes[1].hist(df['y'])
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, sharex=True, sharey=True)

axes[0].hist(df['x'])
axes[0].set_title('x')

axes[1].hist(df['y'])
axes[1].set_title('y')

plt.show()
```

#### *3. pandas.plot API*

```python
### Insert code to plot histogram
```

```python
fig, axes = plt.subplots(1, 2, sharex=True, sharey=True)

df.hist('x', ax=axes[0], grid=False)
df.hist('y', ax=axes[1], grid=False)

plt.show()
```
