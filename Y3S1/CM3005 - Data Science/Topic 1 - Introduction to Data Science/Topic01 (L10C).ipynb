```python
# Data Structure (Dictionary)

person = {"name":"Alice", "age":25, "city":"London"}
print(person)

person["age"]=35
person["profession"]="Engineer"
print(person)
```

```python
print(person.items())
print(person.keys())
print(person.values())
```

```python
for k, v in person.items():
    print(k, v)
```

```python
for k in person.keys():
    print(k)
```

```python
for v in person.values():
    print(v)
```

```python
print(person)
print("city" in person)
print("city" in person.keys())
print()
print("London" in person)
print("London" in person.values())
```

```python
# Data Structure (Set)

unique_vals = {1, 2, 3, 3, 4}
print(unique_vals)
unique_vals.add(55)
print(unique_vals)
unique_vals.remove(3)
print(unique_vals)
```

```python
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1 | set2)  # Union
print(set1 & set2)  # Intersection
print(set1 - set2)  # Difference
print(set2 - set1)  # Difference
```

### Loops

```python
# While loop

count = 0
while count < 3:
    print(count)
    # count = count + 1
    count += 1
```

```python
# For loop

fruits = ["apple", "banana"]
for fruit in fruits:
    print(fruit)
```

```python
for i in range(1,4):  # outer loop
    for j in range(1,5):   # inner loop
        print(i, j, "i*j: ", i*j)
    print()
    
```

```python
for i in range(1,10,2):
    if i == 5:
        break
    print(i)
```

```python
for i in range(1,10,2):
    if i == 5:
        continue
    print(i)
```

### Exception

```python
def divide(a, b):
    return a / b
```

```python
import traceback
try:
    result = divide(10,0)
except Exception as e:
    print("error: ", e)
    print("traceback output: ")
    traceback.print_exc()
```
