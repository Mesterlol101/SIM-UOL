### 2.1.2 Linear Algebra in Practice

#### Introduction to Linear Algebra

- Major branch of mathematics, with significant theoretical background and extensive applications in computer science and data science
- Volume and scope of linear algebra is considerable
- Key concepts in linear algebra include:
    * linear combinations
    * systems of linear equations
    * linear transformations (or linear maps)
- We will focus on scalars, vectors, and matrices and how to use them to solve systems of linear equations

#### Scalars

- In the mathematical sense, scalar are values which have only magnitude, e.g., real numbers or any of their subsets (single number)
- Complex numbers are usually treated as scalars, although in some context they can be considered two-dimensional vectors


#### Vectors

- In 2D and 3D spaces, a vector is a quantity represented by an arrow with both direction and magnitude

    
- In linear algebra, the concept of a vector is generalised for any dimension (n):
    * v = (a<sub>1</sub>, a<sub>2</sub>, ..., a<sub>n</sub>)
- The algebraic notion of a vector corresponds to the array data type, as supported by a number of programming languages

#### Vectors and NumPy

```python
import numpy as np
```

```python
### Insert code to perform vector operations
```

```python
# Scalar multiplication

C = int(5)   # scalar
A = np.array([1,2,3], dtype=int)   # vector
B = C * A
print("A: ", A)
print("C: ", C)
print("B: ", B)
```

```python
print(C + A)
print(A - C)
```

```python
# Vector addition (or subtraction)

X = np.array([3,4,5], dtype=int)   #vector
Y = np.array([5,6,7], dtype=int)   #vecor
Z = np.add(X,Y)

print("X :", X)
print("Y :", Y)
print("Z :", Z)
```

```python
# Dot product

X = np.array([1,2,3], dtype=int)   #vector
Y = np.array([5,10,15], dtype=int)   #vector
D = np.dot(X,Y)

print("X :", X)
print("Y :", Y)
print("D :", D)
```

#### Matrices and NumPy: Determinant

- Every square matrix has a determinant
- Determinant is a single value, which has important mathematical meanings
- One of the key questions is whether the determinant has zero or non-zero value
    * If one row (or column) is a linear combination of another row (or column), determinant is 0
- Calculating a determinant requires multiple additions and multiplications, and in practice it is usually done using a software tool

```python
# Initialise a 3 by 3 matrix (array)
M = np.array([[1,2,3],[2,3,4],[2,5,6]])
```

```python
### Insert code to compute the determinant
```

```python
M
```

```python
D = np.linalg.det(M)
print(M)
print(D)
```

```python
M = np.array([[1,2,3],[2,4,6],[2,5,6]])
D = np.linalg.det(M)
print(M)
print(D)
```

#### Matrices and NumPy: Rank

- Rank of a matrix is the number of linearly independent rows (or columns) in it
- A square matrix with non-zero determinant has rank equals to its dimensions



```python
### Insert code to calculate the rank
```

```python
M = np.array([[1,2,3],[2,3,4],[2,5,6]])
R = np.linalg.matrix_rank(M)
D = np.linalg.det(M)

print(M)
print(R)
print(D)
             
```

```python
M = np.array([[1,2,3],[2,3,4],[4,6,8]])
R = np.linalg.matrix_rank(M)
D = np.linalg.det(M)

print(M)
print(R)
print(D)
```

#### Matrices and NumPy: Trace

- The trace of a matrix is the sum of all elements on the main diagonal.

```python
### Insert code to compute the trace
```

```python
M = np.array([[1,2,3],[2,3,4],[4,6,8]])
R = np.linalg.matrix_rank(M)
D = np.linalg.det(M)
T = np.trace(M)

print(M)
print(R)
print(D)
print(T)
```

#### Matrices and NumPy: Matrix Multiplication

- Key operation in linear algebra
- Result of the multiplication is also a matrix
- The multiplication follows specific rules:
    * It is based on the dot product
    * It requires the number of columns from the first matrix to be equal to the number of rows in the second matrix
    * Multiplying matrices A and B
      

```python
A = np.array([[1,2],[3,4]])
B = np.array([[3,2],[1,4]])
AB = np.matmul(A,B)

print(A)
print(B)
print()
print(AB)
```

#### Matrices and NumPy: Inverse Matrix

- A square matrix with non-zero determinant has an inverse
- In the case of real numbers, where the multiplicative inverse of r is 1/r (e.g., multiplicative inverse of 2 is 0.5)
- The inverse of a matrix M is usually denoted with M<sup>-1</sup>
- M x M<sup>-1</sup> = M<sup>-1</sup> x M = I, where I is the identity matrix
- Multiplying any square matrix with I does not change the matrix
  
  


```python
# Initialise a matrix
M = np.array([[1,2,3],[2,3,4],[2,5,6]])
```

```python
D = np.linalg.det(M)
D
```

```python
### Insert code to compute its inverse
```

```python
M1 = np.linalg.inv(M)
print(M)
print(M1)
```

```python
np.matmul(M, M1)
```

```python
np.matmul(M1, M)
```

```python
np.dot(M1, M)
```
