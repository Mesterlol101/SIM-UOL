# 📉 Gradient Descent for Univariate Linear Regression

This notebook demonstrates **gradient descent**, the foundational optimization algorithm used to train machine learning models.

We'll visualize how gradient descent works for **univariate linear regression** (one input feature) by:
- Fitting a line `y = θ₀ + θ₁·x` to noisy data
- Comparing **Batch Gradient Descent** vs **Stochastic Gradient Descent (SGD)**
- Showing the **loss landscape** in 3D
- Tracking parameter evolution and cost over time

---

## 🧠 What is Gradient Descent?

Gradient descent is an **iterative optimization algorithm** used to minimize a **cost (loss) function** by adjusting model parameters in the direction of **steepest descent**.

For linear regression, we use the **Mean Squared Error (MSE)** cost function:

$$
J(\theta_0, \theta_1) = \frac{1}{2m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right)^2
$$

Where:
- $ h_\theta(x) = \theta_0 + \theta_1 x $ is the hypothesis (predicted value)
- $ m $ is the number of training examples
- $ \theta_0 $ = intercept (bias), $ \theta_1 $ = slope (weight)

The goal: **find $ \theta_0, \theta_1 $ that minimize $ J $**.

## 🔁 How Gradient Descent Updates Parameters

At each iteration, we update the parameters using:

$$
\theta_j := \theta_j - \alpha \cdot \frac{\partial J}{\partial \theta_j}
$$

Where:
- $ \alpha $ = **learning rate** (step size)
- $ \frac{\partial J}{\partial \theta_j} $ = **partial derivative** of cost w.r.t. parameter $ \theta_j $

### 📐 Derivatives for Linear Regression

For **batch gradient descent** (using all data):

$$
\frac{\partial J}{\partial \theta_0} = \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)})
$$
$$
\frac{\partial J}{\partial \theta_1} = \frac{1}{m} \sum_{i=1}^{m} (h_\theta(x^{(i)}) - y^{(i)}) \cdot x^{(i)}
$$

In vectorized form (used in code):
```python
grad = (1/m) * X.T @ (predictions - y)
```
## 🔍 Understanding `grad = (1/m) * X.T` in Linear Regression

This line is part of the gradient computation in linear regression using gradient descent.

### 📘 Variables:
- `X`: Input feature matrix of shape `(m, n)`  
  - `m`: number of training examples  
  - `n`: number of features  
- `X.T`: Transpose of `X`, shape becomes `(n, m)`
- `m`: Scalar representing the number of training examples

### 🧮 Purpose:
This line scales the transpose of the input matrix by the inverse of the number of samples. It's typically used in the gradient calculation for updating model parameters.

### 🧠 Full Gradient Formula:
In linear regression, the gradient of the cost function with respect to the weights `θ` is:



$$
\text{grad} = \frac{1}{m} X^T (X\theta - y)
$$



Your line is a partial step toward this full expression.

## 🧮 Numeric Example of `grad = (1/m) * X.T`

Let’s compute the gradient step using:


$$
\text{grad} = \frac{1}{m} \cdot X^T
$$


### 📘 Given:
- Number of training examples: \( m = 2 \)
- Feature matrix:


$$
X = \begin{bmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{bmatrix}
$$



### 🔁 Step 1: Transpose of \( X \)


$$
X^T = \begin{bmatrix}
1 & 4 \\
2 & 5 \\
3 & 6
\end{bmatrix}
$$



### ✂️ Step 2: Multiply by $$ \frac{1}{m} = \frac{1}{2} $$


$$
\text{grad} = \frac{1}{2} \cdot X^T =
\begin{bmatrix}
0.5 & 2.0 \\
1.0 & 2.5 \\
1.5 & 3.0
\end{bmatrix}
$$



### ✅ Final Result:


$$
\text{grad} =
\begin{bmatrix}
0.5 & 2.0 \\
1.0 & 2.5 \\
1.5 & 3.0
\end{bmatrix}
$$



This matrix is typically used in the full gradient descent update step:


$$
\theta = \theta - \alpha \cdot \left( \frac{1}{m} X^T (X\theta - y) \right)
$$




## 📦 Batch Gradient Descent (Full Dataset)

In **Batch Gradient Descent**, we compute the gradient using **all `m` training examples** at every step.

### ✅ Advantages:
- Smooth, stable convergence
- Guaranteed to converge to global minimum for convex problems

### ❌ Disadvantages:
- Slower per iteration (must process entire dataset)
- Memory-intensive for large datasets

### 💻 Code Implementation:

In our demo, this is implemented in the `precompute_training()` function as:

```python
# Full batch gradient
pred = X_b @ theta          # Shape: (m,)
grad = (1 / m) * X_b.T @ (pred - y)  # Shape: (2,)

## 🎲 Stochastic Gradient Descent (SGD) – One Sample

In **Stochastic Gradient Descent (SGD)**, we compute the gradient using **only one randomly selected training example** at each step.

### ✅ Advantages:
- Very fast per iteration (only one sample)
- Can escape shallow local minima (due to noise)
- Works with streaming/large data

### ❌ Disadvantages:
- Noisy updates → oscillates near minimum
- May never fully converge (but gets close)

### 💻 Code Implementation:

In our demo, SGD is implemented as:

```python
# Stochastic: one random sample
i = np.random.randint(m)        # Pick random index
x_i = X_b[i:i+1]                # Shape: (1, 2) — single example with bias
y_i = y[i]                      # True label for that example
pred_i = x_i @ theta            # Prediction: scalar
grad = x_i.T @ (pred_i - y_i)   # Shape: (2,) — gradient from one sample

## 🔍 Key Differences Summary

| Feature | Batch GD | Stochastic GD |
|--------|--------|--------------|
| **Gradient Computation** | All `m` samples | 1 random sample |
| **Update Equation** | `θ -= α * (1/m) * X.T @ (X@θ - y)` | `θ -= α * (x_i.T @ (x_i@θ - y_i))` |
| **Convergence Path** | Smooth, direct | Noisy, zig-zag |
| **Per-Step Speed** | Slow | Very fast |
| **Memory Usage** | High (stores all data) | Low (one sample) |
| **Final Convergence** | Exact minimum | Oscillates near minimum |

> 💡 **Mini-batch GD** (not shown) uses a small batch (e.g., 32–512 samples) — best of both worlds!

## 💻 Code Structure & Key Components

The interactive demo includes:

### 1. **Data Generation**
- Creates synthetic data: `y = true_slope * x + true_intercept + noise`
- True parameters are hidden; algorithm must discover them

### 2. **Training Modes**
- **Batch GD**: Compute gradient over full dataset
- **Stochastic GD**: Compute gradient over one random sample

### 3. **Visualization Panels**
| Plot | Purpose |
|------|--------|
| **Data & Fit** | Shows current regression line vs true relationship |
| **3D Loss Landscape** | Cost surface over (θ₀, θ₁); shows optimization path |
| **Cost vs θ₀ / θ₁** | 1D slices showing sensitivity of cost to each parameter |
| **Loss vs Iterations** | Tracks convergence of cost over time |

### 4. **Interactive Controls**
- **Learning Rate**: Controls step size (too high → diverge, too low → slow)
- **GD Mode**: Switch between batch and stochastic
- **Generate Data**: Create new random dataset
- **Step / Run All / Stop**: Control training animation
- **3D Rotation**: Adjust view of loss landscape

> 🎯 **Goal**: Build intuition about how gradient descent navigates the loss landscape to find optimal parameters.

- Switch between **Batch** and **Stochastic** mode
- Observe the **3D path** and **loss curve**
- Adjust **learning rate** to see stability differences

````python
# @title
# Gradient Descent Demo – Batch vs Stochastic Mode
# With 3D loss landscape, rotation controls, fixed loss axis, and Stop button

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
import asyncio

plt.style.use('default')

# Global state
state = {
    'X': None,
    'y': None,
    'theta_history': None,
    'cost_history': None,
    'current_step': 0,
    'true_slope': None,
    'true_intercept': None,
    'learning_rate': 0.1,
    'n_samples': 100,
    'cost_grid': None,
    'theta0_grid': None,
    'theta1_grid': None,
    'running': False,
    'view_azim': -60,
    'view_elev': 30,
    'animation_task': None,
    'gd_mode': 'batch'  # 'batch' or 'stochastic'
}

def generate_data():
    np.random.seed()
    n = state['n_samples']
    X = np.random.randn(n)
    true_slope = np.random.uniform(1.0, 3.0)
    true_intercept = np.random.uniform(-1.0, 2.0)
    noise = np.random.randn(n) * 0.5
    y = true_slope * X + true_intercept + noise

    state.update({
        'X': X,
        'y': y,
        'true_slope': true_slope,
        'true_intercept': true_intercept,
        'theta_history': None,
        'cost_history': None,
        'current_step': 0,
        'cost_grid': None
    })

def compute_cost(X_with_bias, y, theta):
    m = len(y)
    pred = X_with_bias @ theta
    return (1 / (2 * m)) * np.sum((pred - y) ** 2)

def precompute_training():
    """Precompute training trajectory for the selected mode."""
    X, y = state['X'], state['y']
    lr = state['learning_rate']
    mode = state['gd_mode']
    X_b = np.column_stack((np.ones(len(X)), X))

    theta = np.array([0.0, 0.0])
    theta_hist = [theta.copy()]
    cost_hist = [compute_cost(X_b, y, theta)]

    m = len(y)
    num_steps = 100 if mode == 'batch' else 200  # SGD needs more steps

    for step in range(num_steps):
        if mode == 'batch':
            # Full batch
            pred = X_b @ theta
            grad = (1 / m) * X_b.T @ (pred - y)
        else:
            # Stochastic: pick one random sample
            i = np.random.randint(m)
            x_i = X_b[i:i+1]  # shape (1, 2)
            y_i = y[i]
            pred_i = x_i @ theta
            grad = x_i.T @ (pred_i - y_i)  # shape (2,)

        theta = theta - lr * grad
        theta_hist.append(theta.copy())
        cost_hist.append(compute_cost(X_b, y, theta))

    state['theta_history'] = np.array(theta_hist)
    state['cost_history'] = np.array(cost_hist)
    state['current_step'] = 0

def precompute_cost_grid():
    if state['cost_grid'] is not None:
        return

    X, y = state['X'], state['y']
    X_b = np.column_stack((np.ones(len(X)), X))

    true_b, true_m = state['true_intercept'], state['true_slope']
    b_range = max(2.0, abs(true_b) * 1.5)
    m_range = max(2.0, abs(true_m) * 1.5)

    theta0_min, theta0_max = true_b - b_range, true_b + b_range
    theta1_min, theta1_max = true_m - m_range, true_m + m_range

    theta0_vals = np.linspace(theta0_min, theta0_max, 40)
    theta1_vals = np.linspace(theta1_min, theta1_max, 40)
    Theta0, Theta1 = np.meshgrid(theta0_vals, theta1_vals)
    Cost = np.zeros_like(Theta0)

    for i in range(len(theta0_vals)):
        for j in range(len(theta1_vals)):
            theta = np.array([Theta0[j, i], Theta1[j, i]])
            Cost[j, i] = compute_cost(X_b, y, theta)

    state.update({
        'theta0_grid': Theta0,
        'theta1_grid': Theta1,
        'cost_grid': Cost,
        'theta0_vals': theta0_vals,
        'theta1_vals': theta1_vals,
        'theta0_min': theta0_min,
        'theta0_max': theta0_max,
        'theta1_min': theta1_min,
        'theta1_max': theta1_max
    })

def plot_1d_cost_slice(ax, X_b, y, current_theta, param_index, true_val, label, color):
    current_val = current_theta[param_index]
    other_val = current_theta[1 - param_index]

    low = min(current_val - 2.0, true_val - 0.5)
    high = max(current_val + 2.0, true_val + 0.5)
    theta_vals = np.linspace(low, high, 100)

    costs = []
    for val in theta_vals:
        test_theta = current_theta.copy()
        test_theta[param_index] = val
        costs.append(compute_cost(X_b, y, test_theta))

    ax.plot(theta_vals, costs, color=color, linewidth=2, alpha=0.8)
    ax.axvline(x=true_val, color='red', linestyle='--', label=f'True {label}', alpha=0.7)
    ax.plot(current_val, compute_cost(X_b, y, current_theta), 'ro', markersize=8, label='Current')

    ax.set_xlabel(label)
    ax.set_ylabel('Cost')
    ax.set_title(f'Cost vs {label}')
    ax.legend()
    ax.grid(True, alpha=0.3)

def update_plot():
    if state['X'] is None:
        return

    X, y = state['X'], state['y']
    step = state['current_step']
    theta_hist = state['theta_history']
    cost_hist = state['cost_history']
    true_m, true_b = state['true_slope'], state['true_intercept']
    X_b = np.column_stack((np.ones(len(X)), X))

    fig = plt.figure(figsize=(16, 15))

    # Row 1, Left: Data & Fit
    ax1 = plt.subplot(3, 2, 1)
    ax1.scatter(X, y, alpha=0.7, label='Data')
    ax1.plot(X, true_m * X + true_b, 'r--', linewidth=2, label='True line')

    if theta_hist is not None and step < len(theta_hist):
        theta = theta_hist[step]
        ax1.plot(X, theta[0] + theta[1] * X, 'g-', linewidth=2, label='Current fit')
        mode_str = "Batch" if state['gd_mode'] == 'batch' else "Stochastic"
        title = f'Data & Fit (Step {step}) – {mode_str}\nθ₀={theta[0]:.4f}, θ₁={theta[1]:.4f}'
    else:
        title = 'Data & Fit'
    ax1.set_title(title)
    ax1.set_xlabel('X')
    ax1.set_ylabel('y')
    ax1.legend()
    ax1.grid(True, alpha=0.3)

    # Row 1, Right: 3D Joint Optimization Path
    ax2 = plt.subplot(3, 2, 2, projection='3d')
    precompute_cost_grid()

    if state['cost_grid'] is not None:
        Theta0 = state['theta0_grid']
        Theta1 = state['theta1_grid']
        Cost = state['cost_grid']

        surf = ax2.plot_surface(Theta0, Theta1, Cost, cmap='viridis', alpha=0.7, linewidth=0, antialiased=True)

        if theta_hist is not None and len(theta_hist) > 1:
            path_theta0 = theta_hist[:step+1, 0]
            path_theta1 = theta_hist[:step+1, 1]
            path_cost = [compute_cost(X_b, y, th) for th in theta_hist[:step+1]]
            ax2.plot(path_theta0, path_theta1, path_cost, 'r-', linewidth=2.5, label='Optimization Path')
            if step < len(theta_hist):
                ax2.scatter([theta_hist[step, 0]], [theta_hist[step, 1]], [path_cost[-1]],
                           color='red', s=80, edgecolor='k', zorder=10, label='Current')

        true_cost = compute_cost(X_b, y, np.array([true_b, true_m]))
        ax2.scatter([true_b], [true_m], [true_cost], color='gold', s=100, edgecolor='k', marker='*', label='True')

        ax2.set_xlim(state['theta0_min'], state['theta0_max'])
        ax2.set_ylim(state['theta1_min'], state['theta1_max'])
        ax2.set_zlim(0, np.max(Cost) * 1.1)

        ax2.set_title('3D Loss Landscape\n(θ₀, θ₁, Cost)')
        ax2.set_xlabel('θ₀ (Intercept)')
        ax2.set_ylabel('θ₁ (Slope)')
        ax2.set_zlabel('Cost')
        ax2.legend(loc='upper left')
        ax2.view_init(elev=state['view_elev'], azim=state['view_azim'])

    # Row 2: Cost vs θ₀
    ax3 = plt.subplot(3, 2, 3)
    if theta_hist is not None and step < len(theta_hist):
        current_theta = theta_hist[step]
        plot_1d_cost_slice(
            ax3, X_b, y, current_theta,
            param_index=0,
            true_val=true_b,
            label='θ₀ (intercept)',
            color='steelblue'
        )
    else:
        ax3.set_title('Cost vs θ₀\n(No training)')
        ax3.grid(True, alpha=0.3)

    # Row 2: Cost vs θ₁
    ax4 = plt.subplot(3, 2, 4)
    if theta_hist is not None and step < len(theta_hist):
        current_theta = theta_hist[step]
        plot_1d_cost_slice(
            ax4, X_b, y, current_theta,
            param_index=1,
            true_val=true_m,
            label='θ₁ (slope)',
            color='seagreen'
        )
    else:
        ax4.set_title('Cost vs θ₁\n(No training)')
        ax4.grid(True, alpha=0.3)

    # Row 3: Loss vs Iterations — FIXED AXIS
    ax5 = plt.subplot(3, 1, 3)
    if cost_hist is not None:
        iterations = np.arange(len(cost_hist))
        y_max = max(cost_hist[0] * 1.1, max(cost_hist) * 1.1, 1.0)
        ax5.set_ylim(0, y_max)
        ax5.set_xlim(0, len(cost_hist) - 1)

        ax5.plot(iterations, cost_hist, 'purple', linewidth=2.5, alpha=0.8, label='Loss')
        if step < len(cost_hist):
            ax5.plot(step, cost_hist[step], 'ro', markersize=8, label=f'Current (Step {step})')
        ax5.set_title('Loss (Cost) Function Over Iterations', fontsize=14, fontweight='bold')
        ax5.set_xlabel('Iteration')
        ax5.set_ylabel('Mean Squared Error (Loss)')
        ax5.legend()
        ax5.grid(True, alpha=0.3)
    else:
        ax5.set_title('Loss Function (No training yet)')
        ax5.grid(True, alpha=0.3)

    plt.tight_layout(pad=3.0)
    plt.show()

# --- UI Widgets ---
lr_widget = widgets.FloatSlider(
    value=0.1,
    min=0.001,
    max=1.0,
    step=0.01,
    description='Learning Rate:',
    readout_format='.3f',
    style={'description_width': 'initial'}
)

mode_dropdown = widgets.Dropdown(
    options=['batch', 'stochastic'],
    value='batch',
    description='GD Mode:',
    style={'description_width': 'initial'}
)

azim_slider = widgets.IntSlider(
    value=-60,
    min=-180,
    max=180,
    step=5,
    description='Azimuth:',
    style={'description_width': 'initial'}
)

elev_slider = widgets.IntSlider(
    value=30,
    min=-90,
    max=90,
    step=5,
    description='Elevation:',
    style={'description_width': 'initial'}
)

btn_generate = widgets.Button(description="Generate New Data", button_style='success')
btn_step = widgets.Button(description="Step", button_style='info')
btn_reset = widgets.Button(description="Reset", button_style='warning')
btn_run_all = widgets.Button(description="Run All", button_style='primary')
btn_stop = widgets.Button(description="Stop", button_style='danger', disabled=True)

output_area = widgets.Output()

# --- Button State Manager ---
def update_button_states():
    running = state['running']
    btn_generate.disabled = running
    btn_step.disabled = running
    btn_reset.disabled = running
    btn_run_all.disabled = running
    btn_stop.disabled = not running
    lr_widget.disabled = running
    mode_dropdown.disabled = running
    azim_slider.disabled = running
    elev_slider.disabled = running

# --- Event Handlers ---
def on_generate(_):
    state['learning_rate'] = lr_widget.value
    state['gd_mode'] = mode_dropdown.value
    generate_data()
    precompute_training()
    update_button_states()
    with output_area:
        clear_output(wait=True)
        update_plot()

def on_step(_):
    if state['theta_history'] is not None:
        state['current_step'] = min(state['current_step'] + 1, len(state['theta_history']) - 1)
        with output_area:
            clear_output(wait=True)
            update_plot()

def on_reset(_):
    state['current_step'] = 0
    update_button_states()
    with output_area:
        clear_output(wait=True)
        update_plot()

def on_lr_change(change):
    state['learning_rate'] = change['new']
    if state['X'] is not None:
        state['gd_mode'] = mode_dropdown.value
        precompute_training()
        state['current_step'] = 0
        with output_area:
            clear_output(wait=True)
            update_plot()

def on_mode_change(change):
    state['gd_mode'] = change['new']
    if state['X'] is not None:
        precompute_training()
        state['current_step'] = 0
        with output_area:
            clear_output(wait=True)
            update_plot()

def on_azim_change(change):
    state['view_azim'] = change['new']
    if not state['running']:
        with output_area:
            clear_output(wait=True)
            update_plot()

def on_elev_change(change):
    state['view_elev'] = change['new']
    if not state['running']:
        with output_area:
            clear_output(wait=True)
            update_plot()

# --- Async Animation ---
async def run_all_animation():
    total_steps = len(state['theta_history']) - 1
    for step in range(state['current_step'] + 1, total_steps + 1):
        if not state['running']:
            break
        state['current_step'] = step

        with output_area:
            clear_output(wait=True)
            update_plot()

        await asyncio.sleep(0.15)

def on_run_all(_):
    if state['theta_history'] is None or state['running']:
        return

    state['running'] = True
    update_button_states()
    state['animation_task'] = asyncio.create_task(run_all_animation())

def on_stop(_):
    state['running'] = False
    update_button_states()
    if state['animation_task'] is not None:
        state['animation_task'].cancel()

# Attach handlers
btn_generate.on_click(on_generate)
btn_step.on_click(on_step)
btn_reset.on_click(on_reset)
btn_run_all.on_click(on_run_all)
btn_stop.on_click(on_stop)
lr_widget.observe(on_lr_change, names='value')
mode_dropdown.observe(on_mode_change, names='value')
azim_slider.observe(on_azim_change, names='value')
elev_slider.observe(on_elev_change, names='value')

# Initialize
generate_data()
precompute_training()
update_button_states()

# Display UI
display(widgets.VBox([
    widgets.HBox([lr_widget, mode_dropdown]),
    widgets.HBox([btn_generate, btn_step, btn_reset, btn_run_all, btn_stop]),
    widgets.HBox([azim_slider, elev_slider]),
    output_area
]))

# Show initial plot
with output_area:
    update_plot()
````

## 🎯 Key Takeaways

### 1. **Learning Rate Matters**
- Too high → overshoots minimum, may diverge
- Too low → converges very slowly
- Just right → smooth, efficient convergence

### 2. **Batch vs Stochastic Trade-offs**
- **Batch GD**:  
  ✅ Smooth path, guaranteed convergence for convex problems  
  ❌ Slow per iteration, memory-intensive

- **Stochastic GD**:  
  ✅ Fast per step, works with large datasets  
  ❌ Noisy updates, may oscillate near minimum

### 3. **The Loss Landscape is Convex**
- For linear regression with MSE, the cost function is **bowl-shaped**
- Only **one global minimum** → gradient descent will always find it (with proper α)

### 4. **Parameter Sensitivity**
- The 1D slices (`Cost vs θ₀`, `Cost vs θ₁`) show how sensitive the model is to each parameter
- Near the minimum, the curve is flat → small changes in θ cause small changes in cost
