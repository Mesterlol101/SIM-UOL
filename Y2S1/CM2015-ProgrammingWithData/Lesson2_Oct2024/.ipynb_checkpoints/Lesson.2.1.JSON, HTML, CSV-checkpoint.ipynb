## JSON – (JavaScript Object Notation)
* A lightweight data-interchange format. 
* Easy for humans to read and write. 
* Easy for machines to parse and generate. 
* A text format that is completely language independent 
* Uses conventions that are familiar to programmers of the C-family of languages, including C, C++, C#, Java, JavaScript, Perl, Python, and many others. 
* An ideal data-interchange language.


##### JSON Data Type
1. *Strings*  "James" "I am James Bond"
2. *Numbers*  38 -10 10.1  1.2e10
3. *Booleans* true false
4. *Arrays*   ["basketball", "swimming", "diving"]
5. *Null*     null
6. *Objects*  {"key": "value"}   {"Age":30}  **{"Hobby": ["basketball", "swimming", "diving"] }**

     "name": "I am James Bond",

Whether you are building a thin client (web application) or thick client (client-server application) at some point you are probably making requests to a web server and need a good data format for responses. 

As of today, there are three major data formats being used to transmit data from a web server to a client: **CSV, XML, and JSON.** 

In order to develop an application with a solid architecture, it’s a good idea to understand the differences between each format and know when to use them.

`````python
import json
def validateJSON(jsonData):
    try:
        json.loads(jsonData)
    except ValueError as err:
        return False
    return True
`````

`````python
InvalidJsonData= """{"name": "jane doe", "salary": 9000, "email": "jane.doe@pynative.com",}"""
isValid = validateJSON(InvalidJsonData)

print("Is the given JSON string is valid?  ", isValid)
`````

*** XML
```xml
<?xml version="1.0" encoding="UTF-8" ?>
<root>
  <name>I am James Bond</name>
  <age>38</age>
  <isAccountant>false</isAccountant>
  <hobby>basketball</hobby>
  <hobby>swimming</hobby>
  <hobby>diving</hobby>
  <children/>
  <friends>
    <name>John</name>
    <age>44</age>
    <isAccountant>true</isAccountant>
  </friends>
</root>
```

#### OR a more detailed version

```xml
<?xml version="1.0" ?>
<root>
	<name type="str">I am James Bond</name>
	<age type="int">38</age>
	<isAccountant type="bool">false</isAccountant>
	<hobby type="list">
		<item type="str">basketball</item>
		<item type="str">swimming</item>
		<item type="str">diving</item>
	</hobby>
	<children type="null"/>
	<friends type="list">
		<item type="dict">
			<name type="str">John</name>
			<age type="int">44</age>
			<isAccountant type="bool">true</isAccountant>
		</item>
	</friends>
</root>
```

XML is a versatile markup language for structuring and sharing data. 
- It uses tags to define elements enclosed in angle brackets, like <element>. 
- Elements can have attributes, such as <element attribute="value">. 
- XML documents must have a single root element containing all other elements. 
```XML
<library>
  <book>
    <title>Sample Title</title>
    <author>John Doe</author>
  </book>
</library>
```
    
- Elements can be nested to represent hierarchical data like the book within the library.
- Empty elements are self-closed, like `<img />.`  See `Children` above
- Comments are enclosed in` <!-- comment -->. `

# CSV Data (Comma-separated values)
* See https://en.wikipedia.org/wiki/Comma-separated_values 
* Basically, it's a text file
* Uses comma to separate values
* Not fully standardized.
* Typically stores tabular data 


`````python
filename="results.csv"
f = open(filename,"r")
for line in f:
    print(line)
f.close()
`````

`````python
filename1="results.csv"
# observed the colon :   still reading ... and will close if the sentence is outside this scope.

with open(filename1) as csvfile:
    for i in csvfile.readlines():
        print(i)
# NO need to close   ---- f.close()
`````

#### Without importing CSV
**import csv**

`````python
filename1="results.csv"
# observed the colon :   still reading ... and will close if the sentence is outside this scope.

with open(filename1, 'r') as csvfile:
    for i in csvfile.readlines():
        print(i.split(','))
`````

The downside of this approach is that it loads the entire file into memory, which can be inefficient for large CSV files.

#### With CSV library

`````python
import csv
with open('results.csv', 'r') as csvfile:
    csv_reader = csv.reader(csvfile)
    for row in csv_reader:
        print(row)
`````

This approach is memory-efficient and simplifies data processing as the csv.reader class takes care of splitting the CSV lines into values based on the specified delimiter (usually a comma).

## Summarizing the differences and similarities between CSV, XML, and JSON:

| Feature          | CSV (Comma-Separated Values) | XML (eXtensible Markup Language) | JSON (JavaScript Object Notation) |
|------------------|------------------------------|----------------------------------|-----------------------------------|
| **Structure**    | Simple, flat structure       | Hierarchical, tree structure     | Hierarchical, tree structure      |
| **Syntax**       | Plain text, comma-separated  | Tags with attributes and values  | Key-value pairs, arrays, objects  |
| **Readability**  | Easy for humans to read      | Less readable due to tags        | Readable, similar to JavaScript   |
| **Data Types**   | Text only                    | Supports various data types      | Supports various data types       |
| **File Size**    | Small                        | Larger due to tags               | Smaller than XML, larger than CSV |
| **Use Cases**    | Simple data export/import    | Complex data representation      | Web APIs, configuration files     |
| **Parsing**      | Easy to parse                | Requires XML parser              | Easy to parse with JSON parsers   |
| **Flexibility**  | Limited                      | Highly flexible                  | Flexible                          |
| **Language Support** | Supported by most languages | Supported by most languages      | Supported by most languages       |

### Similarities:
- **Language Support**: All three formats are widely supported across various programming languages.
- **Data Exchange**: They are commonly used for data exchange between systems.

### Differences:
- **Structure and Syntax**: CSV is flat and simple, while XML and JSON are hierarchical. XML uses tags, whereas JSON uses key-value pairs.
- **Readability**: CSV is the easiest to read, followed by JSON, with XML being the least readable due to its verbose tags.
- **File Size**: CSV files are generally smaller, while XML files are larger due to extensive tagging. JSON files are typically smaller than XML but larger than CSV.
- **Use Cases**: CSV is best for simple data export/import, XML for complex data representation, and JSON for web APIs and configuration files.
