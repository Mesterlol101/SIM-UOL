### Absolute vs Relative Path 


* The absolute path is the full path starting from the root or C drive to a given file or directory.
* The relative path is the path to a given file or directory, with respect to the current working directory `(cwd)`. For example:
  ** Absolute path: C:/users/julian/docs/myfile.txt
  ** If my cwd is C:/users/julian/, then the relative path to `myfile.txt` would be: `docs/myfile.txt`

That is to say, `cwd` + relative path = absolute path.

2. **Code Explanation**:

     - `sys.argv[0]` is the name of the program/script being executed.

   - `print("Number of elements including the name of the program:", len(sys.argv))`:
     - `len(sys.argv)` returns the total number of command-line arguments, including the name of the program.
     - This line prints the total number of elements in `sys.argv`.

   - `print("Number of elements excluding the name of the program:", (len(sys.argv) - 1))`:
     - `(len(sys.argv) - 1)` gives the number of command-line arguments excluding the name of the program.
     - This line prints the number of elements excluding the name of the program.

   - `print("Argument List:", str(sys.argv))`:
     - `str(sys.argv)` converts the `sys.argv` list to a string for easier readability.
     - This line prints the entire list of command-line arguments.

````python
os.chdir(".")
for root, dirs, files in os.walk("."):
    for name in files:
        print(os.path.join(root, name))
    for name in dirs:
        print("this is dir:" + os.path.join(root, name))
        
                               
````


1. **Change Directory**:
   - `os.chdir(".")`
     - This line changes the current working directory to the current directory ("."). Essentially, it ensures the current working directory is set, but since it’s already the current directory, this line doesn’t change much.

2. **Walk Directory Tree**:
   - `for root, dirs, files in os.walk("."):`
     - `os.walk()` is a generator that yields a tuple of `(root, dirs, files)` for each directory it walks through.
     - `root`: The current directory path.
     - `dirs`: A list of sub-directories in `root`.
     - `files`: A list of non-directory files in `root`.

3. **Print File Paths**:
   - `for name in files:`
     - This loop iterates over all the files in the current directory.
   - `print(os.path.join(root, name))`
     - For each file, this prints the full path by joining the `root` directory with the file name.

4. **Print Directory Paths**:
   - `for name in dirs:`
     - This loop iterates over all the sub-directories in the current directory.
   - `print("this is dir:" + os.path.join(root, name))`
     - For each directory, this prints a message indicating it's a directory, along with the full path by joining the `root` directory with the directory name.

### What This Code Does
- The script starts in the current directory.
- It walks through the directory tree from the current directory downwards.
- For every file it encounters, it prints the full path of the file.
- For every directory it encounters, it prints a message along with the full path of the directory.

### Example Output
If the current directory has the following structure:
```

│   └── file1.txt
├── dir2
├── file2.txt
└── file3.txt
```

The output of the script would look something like:
```
./file2.txt
./file3.txt
this is dir:./dir1
this is dir:./dir2
./dir1/file1.txt
```
