# C 

### Error handling in Python using the `try`, `except`, and `finally` blocks.

Here we consider a simple example where we attempt to divide two numbers, and we want to handle the potential division by zero error:

```python
def divide_numbers(numerator, denominator):
    try:
        result = numerator / denominator
    except ZeroDivisionError as e:
        print(f"Error: {e} - Division by zero is not allowed.")
        result = None
    finally:
        print("This block always executes, regardless of whether an exception occurred or not.")
    return result

# Example usage
numerator = 10
denominator = 0

result = divide_numbers(numerator, denominator)

if result is not None:
    print(f"The result of the division is: {result}")
else:
    print("Division was not successful.")
```

Here is the explanation of the code.

- The `try` block contains the code that might raise an exception.
- The `except` block specifies the type of exception to catch (in this case, `ZeroDivisionError`). If the specified exception occurs in the `try` block, the code inside the `except` block will be executed.
- The `finally` block contains code that will always be executed, whether an exception occurs or not.
- If a `ZeroDivisionError` occurs, a message is printed, and the result is set to `None`.
- The `finally` block prints a message indicating that it always executes.
- The function returns the result, which is either the division result or `None` if an exception occurred.

This example demonstrates a basic structure for handling errors and ensuring that certain code (in the `finally` block) is executed regardless of whether an exception occurs or not. 

# D

### An alternative approach for dealing with the error handling scenario, specifically for the division by zero example, is to perform a check before attempting the division. This way, you can avoid triggering an exception altogether, as shown below:

```python
def divide_numbers(numerator, denominator):
    if denominator == 0:
        print("Error: Division by zero is not allowed.")
        result = None
    else:
        result = numerator / denominator
    return result

# Example usage
numerator = 10
denominator = 0

result = divide_numbers(numerator, denominator)

if result is not None:
    print(f"The result of the division is: {result}")
else:
    print("Division was not successful.")
```

so in the code above, the `divide_numbers` function now starts with a check: if the `denominator` is zero, it prints an error message and sets the result to `None`.  If the `denominator` is non-zero, it proceeds with the division and returns the result.

````python
from graphviz import Digraph

# Create a new Digraph
process_flow = Digraph('process_flow', format='png')

# Add nodes
process_flow.node('start', shape='ellipse', label='Start')
process_flow.node('read_file', label='Read file')
process_flow.node('split_words', label='Split words')
process_flow.node('initialize_counter', label='Initialize counter')
process_flow.node('for_loop', label='For each word')
process_flow.node('check_existence', label='Check existence')
process_flow.node('increment_count', label='Increment count')
process_flow.node('add_to_counter', label='Add to counter')
process_flow.node('print_counts', label='Print counts')
process_flow.node('cleanup', label='Cleanup')
process_flow.node('stop', shape='ellipse', label='Stop')

# Add edges
process_flow.edge('start', 'read_file')
process_flow.edge('read_file', 'split_words')
process_flow.edge('split_words', 'initialize_counter')
process_flow.edge('initialize_counter', 'for_loop')
process_flow.edge('for_loop', 'check_existence')
process_flow.edge('check_existence', 'increment_count', label='Yes')
process_flow.edge('check_existence', 'add_to_counter', label='No')
process_flow.edge('increment_count', 'for_loop')
process_flow.edge('add_to_counter', 'for_loop')
process_flow.edge('for_loop', 'print_counts', constraint='false')  # Loops back
process_flow.edge('print_counts', 'cleanup')
process_flow.edge('cleanup', 'stop')

# Render the diagram
process_flow.render('process_flow', view=True)  # Saves as process_flow.png
````

### Provide an example of counting words by utilising the power of an external library.

````python
import nltk
from nltk.tokenize import word_tokenize
from collections import Counter

# Sample text
text = "Should I share memes, or should I not share memes, that's what I'm pondering:\
Is it better to deal with the mean comments and criticisms online,\
Or fight against the never-ending stream of cute cat videos,\
And by sharing, put an end to them. To laugh, or to sleep;\
No more; and by sleeping, say we end\
The heartache and the countless annoying things\
That come with using social media, it's something\
I really hope for. To laugh, to sleep"

# Tokenize the text into words
words = word_tokenize(text)

# Convert all words to lowercase
words = [word.lower() for word in words]

# Count the occurrences of each word
word_counts = Counter(words)

# Print the word counts
print("Word Counts:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
````

### Provide three examples of how you might standardise the sort of data you see in Video 3.07

Here are some examples of how you might standardize the sort of data:

1. **Convert Words to Lowercase**: Ensure that all words are in lowercase so that "Word" and "word" are treated the same way during analysis.

2. **Remove Punctuation**: Strip punctuation marks from the words so that "word," and "word" are counted as the same word.

3. **Remove Stop Words**: Remove common words like "and," "the," "is," etc., which do not contribute much to the analysis of word frequencies.

6. **Remove Numerical Values**: If the data contains numerical values, you might want to remove them if they are not relevant to your analysis of word frequencies.
```python

```

These standardization techniques help ensure that the data is clean, consistent, and ready for analysis. They help in avoiding redundancy and ensuring that similar variations of words are treated the same way.

````python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from collections import Counter
import re

# Sample text
text = "Should I share memes, or should I not share memes, that's what I'm pondering:\
Is it better to deal with the mean comments and criticisms online,\
Or fight against the never-ending stream of cute cat videos,\
And by sharing, put an end to them. To laugh, or to sleep;\
No more; and by sleeping, say we end\
The heartache and the countless annoying things\
That come with using social media, it's something\
I really hope for. To laugh, to sleep"

# Tokenize the text into words
words = word_tokenize(text.lower())

# Remove punctuation
words_no_punct = [re.sub(r'[^\w\s]', '', word) for word in words]

# Remove stop words
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words_no_punct if word not in stop_words]

# Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

# Remove numerical values
words_no_numbers = [re.sub(r'\d+', '', word) for word in stemmed_words]

# Count the occurrences of each word
word_counts = Counter(words_no_numbers)

# Print the word counts
print("Word Counts:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
````
